1. Harga produk distibusi miring ke kanan, berarti mayoritas dijual dengan harga terjangkau. Akan tetapi rata-rata tertarik karena ada beberapa produk yang harganya sangat mahal
2. Banyak terjual, disini ada perbedaan antar toko yang cukup ekstrem dimana sebagian besar toko memiliki penjualan standar dan ada toko yang sudah besar endominasi penjualan
3. Rating produk, distribusi miring ke kiri, tandanya mayoritas pembeli puas dengan produk tetapi rata-rata menurun karena ada review kurang bagus dari pembeli.

dari ketiga data diatas tidak ada yang berdistribusi normal tapi sesuai dengan instruksi sebelumnya makan outlier akan dibiarkan. Kita akan anggap distribusi normal dan lanjut ke perhitungan analisis omset per bulan dari listing produk dengan CI 95%

```python
df['omset'] = df['harga_produk']*df['banyak_terjual'] #membuat kolom baru untuk omset

mean_omset = df['omset'].mean() 
std_omset = df['omset'].std()
N=len(df['omset'])

lower_ci, upper_ci = stats.norm.interval(0.95, loc = mean_omset, scale= std_omset/np.sqrt(N))

print (f"Rata-rata omset: Rp {mean_omset}")
print (f"Lower CI: Rp {lower_ci}")
print (f"Upper CI: Rp {upper_ci}")
```

Hasil dari CI dengan keyakinan 95% menunjukkan bahwa, rata-rata omset populasi listing produk seblak berada di antara 10,25jt hingga 115,3 jt per bulan, sehingga target kita untuk mendapatkan 6,7jt per bulan dapat terpenuhi. Akan tetapi kita perlu membandingkan apakah bahan baku di kota yang berbeda berdampak terhadap penjualan

```python
df['omset'].median()
(df['omset'] >= 6700000).mean()
```

Pertama kita perlu tau jumlah listing per kota, membuat kolom baru berdasarkan wilayah dan menentukan harga per wilayahnya

```python
print(df['kota_toko'].value_counts())
print()
```

Karena sudah tau kota apa saja mari kita grouping satu per satu wilayahnya

```python
Jabodetabek = ['Jakarta Timur', 'Jakarta Selatan', 'Kab. Tangerang', 'Tangerang', 'Depok', 'Kab. Bekasi']
Luar_Jabodetabek = ['Bandung', 'Kab. Bandung',  'Kab.Ciamis', 'Kab. Garut', 'Kab. Tasikmalaya', 'Malang', 'Surabaya', 'Tasikmalaya']

harga_jabodetabek = df[df['kota_toko'].isin(Jabodetabek)]['harga_produk']
harga_luar_jabodetabek = df[df['kota_toko'].isin(Luar_Jabodetabek)]['harga_produk']

print (f'Jumlah Produk Jabodetabek: {len(harga_jabodetabek)}')
print (f'Jumlah Produk Luar Jabodetabek: {len(harga_luar_jabodetabek)}')
print (f'Rata-Rata Harga Produk Jabodetabek: Rp {harga_jabodetabek.mean()}')
print (f'Rata-Rata Harga Produk Luar Jabodetabek: {harga_luar_jabodetabek.mean()}')
```

Hipotesis dan Uji

- H0: rata-rata harga produk di Jabodetabek sama dengan di luar Jabodetabek
- H1: rata-rata harga produk di Jabodetabek berbeda dengan di luar Jabodetabek

menggunakan uji t test sample karena kita sedang membandingkan dua kelompok


```python
t_stat, p_value = stats.ttest_ind(harga_jabodetabek, harga_luar_jabodetabek)

print (f'T-Statistic : {t_stat}')
print (f'P_value : {p_value}')

alpha = 0.05
if p_value < alpha:
    print ("Kesimpulan : Tolak H0. Terdapat perbedaan rata-rata harga produk yang signifikan antara toko wilayah Jabodetabek dan Luar Jabodetabek")
else:
    print ("Kesimpulan: Gagal menolak H0. Tidak terdapat perbedaan rata-rata harga produk yang signifikan antara toko wilayah Jabodetabek dan Luar Jabodetabek")
```

Temuan diatas menunjukkan bahwa adanya kemungkinan karena seblak adalah produk instan sehingga bahan baku tidak berpengaruh terhadap penjualan, karena tidak tergantung wilayahnya. Karena wilayah tidak berpengaruh maka ada pertanyaan mengenai korelasi antara produk itu sendiri dengan ratingnya. Apakah berpengaruh dengan cara melakukan uji korelasi spearman dan kendall-tau karena data tidak berdistribusi normal.

H0: Tidak terdapat korelasi yang signifikan antara harga produk dengan rating produk
H1: Terdapat korelasi yang signifikan antara harga produk dengan rating produk

```python
corr_spearman, pval_spearman = stats.spearmanr(df['harga_produk'], df['rating_produk'])
corr_kendall, pval_kendall = stats.kendalltau(df['harga_produk'], df['rating_produk'])

print(f'Korelasi Spearman: {corr_spearman} (p-value: {pval_spearman})')
print(f'Korelasi Kendall-Tau: {corr_kendall} (p-value: {pval_kendall})')

alpha = 0.05
if pval_spearman < alpha:
    print ("Kesimpulan: Tolak H0. Terdapat hubungan yang signifikan secara statistik antara harga dan rating produk")
else:
    print ("Kesimpulan: Gagal menolak H0. Tidak terdapat hubungan yang signifikan secara statistik antara harga dan rating produk")
```

Nilai korelasi Spearman = 0,26 menunjukan bahwa adanya hubungan positif antara harga dengan rating produk dengan p-value = 0,048 yang artinya terdapat hubungan yang signifikan secara statistik diantara harga dan rating produk.

Kesimpulan
p-value ini memberikan bukti bahwa secara analisis memang betul harga produk memiliki keterkaitan yang tinggi dengan ulasan/review buyer
