```python
import pandas as pd
import numpy as np

# 1. DATA LOADING
df = pd.read_csv('portugal_listinigs.csv', low_memory=False)

# Menampilkan 5 baris pertama data
df.head()
```

```python
# 2.EKSPLORASI DATA DASAR
# Memeriksa tipe data dan jumlah non-null dari tiap kolom
df.info()

# Memeriksa ukuran dataset (jumlah baris dan kolom)
print("Dimensi dataset (Baris, Kolom):", df.shape)

# Mengecek jumlah missing value per kolom
print("\nJumlah Missing Values:")
print(df.isnull().sum())

# Menampilkan ringkasan statistik untuk kolom numerik
df.describe()
```

```python
# 3. DATA CLEANING
# Menghapus data duplikat
print(f"Jumlah_Baris_Awal: {len(df)}")
df = df.drop_duplicates()
print(f"Handling_Jumlah_Baris: {len(df)}")

# Menghapus baris yang nilai Price atau TotalArea-nya kosong (karena ini variabel utama)
df = df.dropna(subset=['Price', 'TotalArea'])

# Mengisi missing value pada kolom kategori penting (EnergyCertificate) dengan 'Unknown'
if 'EnergyCertificate' in df.columns:
    df['EnergyCertificate'] = df['EnergyCertificate'].fillna('Unknown')

# Verifikasi sisa missing values pada kolom utama
print("Missing_Values_Utama:")
print(df[['Price', 'TotalArea', 'EnergyCertificate']].isnull().sum())
```

```python
# 4. DATA HANDLING
# Membuat kolom baru: Harga per meter persegi (price_per_sqm)
df['price_per_sqm'] = df['Price'] / df['TotalArea']

# Menampilkan pratinjau kolom baru
df[['Price', 'TotalArea', 'price_per_sqm']].head()
```

```python
# 4. DATA SAVING AND COMPACTING
# Menyimpan data bersih 
df.to_csv('portugal_listinigs_clean.csv', index=False)

#Saring Data agar lebih compact
df[['Price']].to_csv('col_price.csv', index=True)
df[['District']].to_csv('col_district.csv', index=True)
df[['City']].to_csv('col_city.csv', index=True)
df[['Town']].to_csv('col_town.csv', index=True)
df[['Type']].to_csv('col_type.csv', index=True)
df[['EnergyCertificate']].to_csv('col_energy.csv', index=True)
df[['TotalArea']].to_csv('col_totalarea.csv', index=True)
df[['NumberOfBathrooms']].to_csv('col_bathrooms.csv', index=True)
df[['Parking']].to_csv('col_parking.csv', index=True)
df[['Elevator']].to_csv('col_elevator.csv', index=True)
df[['price_per_sqm']].to_csv('col_price_per_sqm.csv', index=True)
```
